```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

data = load_breast_cancer()
X, Y = data.data, data.target

x_train, x_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.3, stratify=Y, random_state=42
)

print('Original Dimension: ', x_train.shape)

scaler = StandardScaler()
x_train_scaler = scaler.fit_transform(x_train)
x_test_scaler = scaler.transform(x_test)
```

```python
best_k = 0
best_acc = 0

for k in range(1, 21):
    knn = KNeighborsClassifier(n_neighbors=k, metric='euclidean')
    knn.fit(x_train_scaler, y_train)
    y_pred = knn.predict(x_test_scaler)
    acc = accuracy_score(y_test, y_pred)
    print(f'With k: {k:>2}, Accuracy: {acc:>.4f}')

    if acc > best_acc:
        best_acc = acc
        best_k = k

print(f'\nBest k value: {best_k}, Accuracy: {best_acc:.4f}') 
```

```python
pca = PCA(n_components=10)
x_train_pca = pca.fit_transform(x_train_scaler)
x_test_pca = pca.transform(x_test_scaler)

print('Reduced dimensions: ', x_train_pca.shape)

model_pca = KNeighborsClassifier(n_neighbors=best_k, metric='euclidean')
model_pca.fit(x_train_pca, y_train)
y_pred_pca = model_pca.predict(x_test_pca)
acc_pca = accuracy_score(y_test, y_pred_pca)

print(f'Accuracy After PCA:  {acc_pca:.4f}')
```
