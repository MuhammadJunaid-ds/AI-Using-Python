```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.cluster import AgglomerativeClustering
```

```python
data = load_iris()
X = pd.DataFrame(data.data, columns=data.feature_names)

print('Dataset Shape: ', X.shape)
print('\nFirst 5 rows of Dataset: \n', X.head())
```

```python
hc = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)
```

```python
cluster_labels = hc.fit_predict(X)

X['Clusters'] = cluster_labels

print('Cluster Distribution: ')
print(X['Clusters'].value_counts())
```

```python
print('Sample Cluster labels: ')
print(X.head())
```

```python
from sklearn.metrics import accuracy_score

print('Accuracy: ',accuracy_score(data.target, cluster_labels))
```
